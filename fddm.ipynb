#### Traitement de la variable cible

```python
print(df['FlagImpaye'].unique())
```

```python
# Nettoyage de la colonne 'FlagImpaye' en remplaçant les mauvaises valeurs
df['FlagImpaye'] = df['FlagImpaye'].replace({'FlagImpaye': None, '0': 0, '1': 1})

# Remplacement des NaN par 0 (ou 1 selon votre logique) pour gérer les valeurs manquantes
df['FlagImpaye'].fillna(0, inplace=True)

# Conversion de la colonne en entier
df['FlagImpaye'] = df['FlagImpaye'].astype('int')
```

```python
# Vérification après nettoyage
print(df['FlagImpaye'].unique())
```

```python
# vérifier valeurs manquantes
print(df.isnull().sum())
```

#### Train / Test

```python
# Séparation des données en ensembles d'entraînement et de test
train_data = df[(df['DateTransaction'] >= '2017-02-01') & (df['DateTransaction'] <= '2017-08-31')]
test_data = df[(df['DateTransaction'] >= '2017-09-01') & (df['DateTransaction'] <= '2017-11-30')]
```

```python
# Encodage de la variable 'CodeDecision'
df['CodeDecision'] = df['CodeDecision'].astype('category').cat.codes
```

```python
print(train_data.columns)
```

```python
# Séparation de X et y
X_train = train_data.drop(columns=['FlagImpaye', 'DateTransaction', 'IDAvisAutorisationCheque', 'ZIBZIN'])
y_train = train_data['FlagImpaye']

# Vérification des types
print(X_train.dtypes)
```

```python
from imblearn.over_sampling import SMOTE

# Remplacer les virgules par des points et convertir en float
X_train = X_train.replace({',': '.'}, regex=True)

# Convertir toutes les colonnes en numériques (erreur sera ignorée pour les colonnes non numériques)
X_train = X_train.apply(pd.to_numeric, errors='coerce')

# Vérification des types après conversion
print(X_train.dtypes)

# Application de SMOTE pour équilibrer les classes
smote = SMOTE(random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)
```

```python
# Vérification du nombre d'exemples après SMOTE
print(pd.Series(y_train_res).value_counts())
```

```python
# Normalisation/Standardisation des données
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_res_scaled = scaler.fit_transform(X_train_res)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

# Création et entraînement d'un modèle RandomForest
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train_res_scaled, y_train_res)

# Prédiction sur les données de test
X_test = test_data.drop(columns=['FlagImpaye', 'DateTransaction', 'IDAvisAutorisAtionCheque', 'ZIBZIN'])
y_test = test_data['FlagImpaye']
X_test_scaled = scaler.transform(X_test)

y_pred = rf_model.predict(X_test_scaled)

# Évaluation du modèle
print(classification_report(y_test, y_pred))
```
